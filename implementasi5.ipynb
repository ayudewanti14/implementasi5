```python
!pip -q install plotly ipywidgets

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import ipywidgets as widgets

from IPython.display import display, clear_output, HTML

print("Library berhasil disiapkan!")
```

```python
np.random.seed(42)

n = 150

df = pd.DataFrame({
    "ID_Pengguna": [f"U{i:03d}" for i in range(1, n + 1)],
    "Usia": np.random.randint(18, 51, n),
    "Perangkat": np.random.choice(
        ["Mobile", "Desktop", "Tablet"], n,
        p=[0.60, 0.30, 0.10]
    ),
    "Pengalaman": np.random.choice(
        ["Pemula", "Menengah", "Mahir"], n
    ),
    "Tugas_Berhasil": np.random.choice(
        ["Berhasil", "Gagal"], n, p=[0.82, 0.18]
    ),
    "Waktu_Tugas_Detik": np.random.randint(15, 181, n),
    "Skor_Kepuasan": np.random.randint(1, 6, n),
    "Skor_SUS": np.random.randint(35, 101, n),
    "Jenis_Tugas": np.random.choice(
        ["Login", "Pencarian", "Pengisian Formulir",
         "Navigasi", "Unduh Laporan"], n
    )
})

# Mengelompokkan pengguna berdasarkan usia
df["Persona"] = pd.cut(
    df["Usia"],
    bins=[17, 25, 35, 50],
    labels=["Pengguna Muda", "Dewasa", "Dewasa Senior"]
)

display(df.head(10))
```

```python
jumlah_pengguna = len(df)

tingkat_keberhasilan = (
    df["Tugas_Berhasil"].eq("Berhasil").mean() * 100
)

rata_waktu = df["Waktu_Tugas_Detik"].mean()
rata_kepuasan = df["Skor_Kepuasan"].mean()
rata_sus = df["Skor_SUS"].mean()

print("===== HASIL EVALUASI USABILITY =====")
print(f"Jumlah pengguna       : {jumlah_pengguna}")
print(f"Tingkat keberhasilan  : {tingkat_keberhasilan:.2f}%")
print(f"Rata-rata waktu tugas : {rata_waktu:.2f} detik")
print(f"Rata-rata kepuasan    : {rata_kepuasan:.2f} dari 5")
print(f"Rata-rata skor SUS    : {rata_sus:.2f} dari 100")
```

```python
hasil_tugas = (
    df["Tugas_Berhasil"]
    .value_counts()
    .rename_axis("Status")
    .reset_index(name="Jumlah")
)

fig = px.bar(
    hasil_tugas,
    x="Status",
    y="Jumlah",
    color="Status",
    text="Jumlah",
    title="Keberhasilan Pengguna Menyelesaikan Tugas",
    labels={
        "Status": "Status Tugas",
        "Jumlah": "Jumlah Pengguna"
    },
    color_discrete_map={
        "Berhasil": "#16866B",
        "Gagal": "#D9534F"
    }
)

fig.update_layout(
    template="plotly_white",
    showlegend=False
)

fig.show()
```

```python
waktu_tugas = (
    df.groupby("Jenis_Tugas", as_index=False)
    ["Waktu_Tugas_Detik"].mean()
    .sort_values("Waktu_Tugas_Detik", ascending=True)
)

fig = px.bar(
    waktu_tugas,
    x="Waktu_Tugas_Detik",
    y="Jenis_Tugas",
    orientation="h",
    text_auto=".1f",
    title="Rata-rata Waktu Penyelesaian Setiap Tugas",
    labels={
        "Waktu_Tugas_Detik": "Waktu (detik)",
        "Jenis_Tugas": "Jenis Tugas"
    }
)

fig.update_layout(
    template="plotly_white",
    yaxis={"categoryorder": "total ascending"}
)

fig.show()
```

```python
persona_data = (
    df.groupby("Pengalaman", observed=True)
    .size()
    .reset_index(name="Jumlah_Pengguna")
)

fig = px.pie(
    persona_data,
    names="Pengalaman",
    values="Jumlah_Pengguna",
    title="Distribusi Persona Berdasarkan Pengalaman",
    hole=0.4
)

fig.update_traces(textposition="inside", textinfo="percent+label")
fig.show()
```

```python
kepuasan_perangkat = (
    df.groupby("Perangkat", as_index=False)
    .agg(
        Rata_Kepuasan=("Skor_Kepuasan", "mean"),
        Rata_SUS=("Skor_SUS", "mean"),
        Jumlah_Pengguna=("ID_Pengguna", "count")
    )
)

fig = px.bar(
    kepuasan_perangkat,
    x="Perangkat",
    y="Rata_Kepuasan",
    text_auto=".2f",
    title="Rata-rata Kepuasan Pengguna Berdasarkan Perangkat",
    labels={
        "Perangkat": "Perangkat",
        "Rata_Kepuasan": "Skor Kepuasan (1–5)"
    }
)

fig.update_yaxes(range=[0, 5])
fig.update_layout(template="plotly_white")
fig.show()

display(kepuasan_perangkat)
```

```python
filter_perangkat = widgets.Dropdown(
    options=["Semua"] + sorted(df["Perangkat"].unique().tolist()),
    value="Semua",
    description="Perangkat:"
)

filter_pengalaman = widgets.Dropdown(
    options=["Semua"] + sorted(df["Pengalaman"].unique().tolist()),
    value="Semua",
    description="Pengalaman:"
)

output = widgets.Output()

def render_dashboard(*args):
    with output:
        clear_output(wait=True)

        data = df.copy()

        if filter_perangkat.value != "Semua":
            data = data[
                data["Perangkat"] == filter_perangkat.value
            ]

        if filter_pengalaman.value != "Semua":
            data = data[
                data["Pengalaman"] == filter_pengalaman.value
            ]

        if data.empty:
            print("Tidak ada data yang sesuai dengan filter.")
            return

        sukses = (
            data["Tugas_Berhasil"].eq("Berhasil").mean() * 100
        )
        waktu = data["Waktu_Tugas_Detik"].mean()
        kepuasan = data["Skor_Kepuasan"].mean()
        sus = data["Skor_SUS"].mean()

        display(HTML(f"""
        <div style="
            padding:18px;
            background:#172554;
            color:white;
            border-radius:10px;
            font-family:Arial">
            <h2>Dashboard Evaluasi Usability</h2>
            <p>Analisis interaksi dan pengalaman pengguna</p>
        </div>
        """))

        ringkasan = pd.DataFrame({
            "Indikator": [
                "Jumlah Pengguna",
                "Keberhasilan Tugas",
                "Rata-rata Waktu",
                "Kepuasan Pengguna",
                "Skor SUS Simulasi"
            ],
            "Nilai": [
                f"{len(data)} orang",
                f"{sukses:.1f}%",
                f"{waktu:.1f} detik",
                f"{kepuasan:.2f} / 5",
                f"{sus:.1f} / 100"
            ]
        })

        display(ringkasan)

        status = (
            data["Tugas_Berhasil"]
            .value_counts()
            .rename_axis("Status")
            .reset_index(name="Jumlah")
        )

        fig1 = px.bar(
            status,
            x="Status",
            y="Jumlah",
            color="Status",
            title="Status Keberhasilan Tugas",
            text="Jumlah",
            color_discrete_map={
                "Berhasil": "#16866B",
                "Gagal": "#D9534F"
            }
        )
        fig1.update_layout(
            template="plotly_white",
            showlegend=False
        )
        fig1.show()

        tugas = (
            data.groupby("Jenis_Tugas", as_index=False)
            ["Waktu_Tugas_Detik"].mean()
            .sort_values("Waktu_Tugas_Detik")
        )

        fig2 = px.bar(
            tugas,
            x="Waktu_Tugas_Detik",
            y="Jenis_Tugas",
            orientation="h",
            title="Waktu Rata-rata per Tugas",
            labels={
                "Waktu_Tugas_Detik": "Waktu (detik)",
                "Jenis_Tugas": "Jenis Tugas"
            }
        )
        fig2.update_layout(template="plotly_white")
        fig2.show()

        kepuasan = (
            data.groupby("Perangkat", as_index=False)
            ["Skor_Kepuasan"].mean()
        )

        fig3 = px.bar(
            kepuasan,
            x="Perangkat",
            y="Skor_Kepuasan",
            title="Kepuasan Berdasarkan Perangkat",
            text_auto=".2f"
        )
        fig3.update_yaxes(range=[0, 5])
        fig3.update_layout(template="plotly_white")
        fig3.show()

        display(HTML("""
        <p><b>Catatan:</b> Periksa tugas dengan waktu pengerjaan
        tinggi atau tingkat kegagalan besar sebagai kandidat
        untuk evaluasi usability lebih lanjut.</p>
        """))

filter_perangkat.observe(render_dashboard, names="value")
filter_pengalaman.observe(render_dashboard, names="value")

display(widgets.HBox([filter_perangkat, filter_pengalaman]))
display(output)

render_dashboard()
```
